```python
percepts = []
table = {
    "Red": "Stop",
    "Yellow": "Slow Down",
    "Green": "Go"
}

def lookup_table(percept,table):
    action = table.get(tuple(percept))
    return action

def table_driven_agent(percept):
    percepts.append(percept)
    action = lookup_table(percepts,table)
    return action

print(percepts, '\t\t', table_driven_agent("Red"))
print(percepts, '\t\t', table_driven_agent("Yellow"))
print(percepts, '\t\t', table_driven_agent("Green"))
print(percepts, '\t\t', table_driven_agent("Blue"))

```
