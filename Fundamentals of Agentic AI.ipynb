```python
table = {
    "Red": "Stop",
    "Yellow": "Slow Down",
    "Green": "Go"
}

percept_sequence = ["Red", "Green", "Yellow", "Red"]

print("Percept\t\tAction")
for percept in percept_sequence:
    action = table.get(percept)
    print(f"{percept}\t\t{action}")
```

```python
def reflex_agent(percept):
    if percept == "Dark":
        return "Turn Light On"
    elif percept == "Dim":
        return "Leave Light As Is"
    elif percept == "Bright":
        return "Turn Light Off"
    else:
        return "No Action"

def run_reflex_agent():
    for percept in percept_sequence:
        action = reflex_agent(percept)
        print(f"Percept: {percept}, Action: {action}")

percept_sequence = ["Dark", "Dim", "Bright", "Dark"]
run_reflex_agent()
```

```python
model = {'Room1': None, 'Room2': None}
heater_on_count = {'Room1': 0, 'Room2': 0}

def update_model(location, status):
    model[location] = status

def temperature_agent(room, reading):
    update_model(room, reading)
    
    if reading == "Cold":
        heater_on_count[room] += 1
        return "Turn Heater ON"
    elif reading == "Hot":
        return "Turn Heater OFF"
    elif model['Room1'] == 'Normal' and model['Room2'] == 'Normal':
        return "All Rooms Comfortable- Idle"
    else:
        return "Move to Other Room to Check"

percept_sequence = [
    ("Room1", "Cold"),
    ("Room1", "Normal"),
    ("Room2", "Hot"),
    ("Room2", "Normal")
]

print(f"{'Room'}         {'Reading'}            {'Action'}                                {'Model'}")

for room, reading in percept_sequence:
    action = temperature_agent(room, reading)
    print(f"{room}         {reading}         {action}                          {model}")

print("\nHeater ON count:", heater_on_count)
```
